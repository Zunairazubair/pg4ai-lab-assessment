```python
stu=[]
for i in range(1,6):
    print("enter details of sudent\n")
    name=input("enter name ")
    r_n=input("enter roll ")
    python=int(input("enter marks of python"))
    ai=int(input("enter marks of AI"))
    maths=int(input("enter marks of maths"))
    student={
        "name":name,
        "roll_no":r_n,
        "sub_marks":{
            "pyt":python,
            "AI":ai,
            "mathematics":maths,
        }
    }
    stu.append(student)
def calculation(student):
    p= student["sub_marks"]["pyt"]
    a= student["sub_marks"]["AI"]
    m= student["sub_marks"]["mathematics"]
    total = p+ a+ m
    perc=(total/300)*100
    if(perc>=80 and perc<=100):
        grade="A"
    elif(perc<80 and perc >=70):
        grade="B"
    elif(perc<70 and perc >=60):
        grade="c"
    elif(perc<60 and perc >=50):
        grade="D"
    elif(perc<50 and perc >=0):
        grade="F"
    else:
        print("invalid marks ")
    print("name = ",student["name"])
    print("total marks = ",total)
    print("percentage = ",perc)
    print("grade = ",grade)
    student["total"] = total
    student["percentage"] = perc
    student["grade"] = grade
    f=open("result.txt","a")
    f.write("Name = " + student["name"] + "\n")
    f.write("Total marks = " + str(total) + "\n")
    f.write("Percentage = " + str(perc) + "%\n")
    f.write("Grade = " + grade + "\n")
    f.close 
for student in stu:
    calculation(student)
highest = stu[0]
for student in stu:
    if student["percentage"] > highest["percentage"]:
        highest = student
print("\n==== STUDENT WITH HIGHEST PERCENTAGE =====")
print("Name =", highest["name"])
print("Roll No =", highest["roll_no"])
print("Percentage =", highest["percentage"])
print("Grade =", highest["grade"])
```

```python
product=[]
for i in range(1,6):
    pname=input("enter name: ")
    pprice=int(input("enter price : "))
    quan=int(input("enter quantity:"))
    prod={
    "name":pname,
    "price":pprice,
    "qunatity":quan
    }
    product.append(prod)
def calculation(prod):
    p=prod["price"]
    q=prod["qunatity"]
    prod_price=p * q
    prod["total_price"]= prod_price    
    return prod_price
bill=0;
for prod in product:
    bill+=calculation(prod)
if(bill>=10000):
    dis=bill*0.10
else:
    dis=0
total_bill=bill-dis
print("total bill = ",total_bill)
highest=product[0]
for prod in product:
    if(prod["price"]>highest["price"]):
       highest=prod
print("most expensive product ")
print("name=",highest["name"])
print("price=",highest["price"])
print("qunatity=",highest["qunatity"])
print("total price=",highest["total_price"])
for prod in product:
    f=open("bill.txt","a")
    f.write("name = :" + prod["name"] + "\n")
    f.write("price = :" + str(prod["price"]) + "\n")
    f.write("quantity = :" + str(prod["qunatity"]) + "\n")
    f.write("total price = :" + str(prod["total_price"]) + "\n")
    f.close()
f=open("bill.txt","a")
f.write("total bill = :" + str(total_bill) + "\n")
f.close()
```
